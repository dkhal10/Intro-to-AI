Now let's classify the point as either M or F.  We'll do this by setting k = 1 and using the Euclidean distance.  We'll define that as:

```python
def euclideanDistance(instance1, instance2, length):
   distance = 0
   for x in range(length):
         #print ('x is ' , x)
         num1=float(instance1[x])
         num2=float(instance2[x])
         distance = distance + pow(num1-num2, 2)
   return math.sqrt(distance)
```

Questions:

1. What is the label for point if k=1?
2. What is the label for point if k=3?
3. Would the result be different if we used a different distance function like Hamming or Manhattan?

```python
#Add code with functions that implement Hamming and Manhattan distances and test
#what the label will be for k=1 and k=3 for all possibilities
#(6 total: 2 Euclidean, 2 Hamming and 2 Manhattan)
def manhattanDistance(one, two, length):
    distance = 0
    for each in range(length):
        distance += abs(float(one[each]) - float(two[each]))
    return distance
def hammingDistance(one, two, length):
    distance = 0
    for each in range(length):
        if one[each] != two[each]:
            distance += 1
    return distance

def knn(point, dataset, k, distanceFunction):
    distances = []
    for each in dataset:
        # d = distanceFunction(point, each, 29)
        d = distanceFunction(point, each, len(point) - 1)
        distances.append([d, each[-1]])
    distances.sort()
    m = 0
    f = 0
    for each in range(k):
        if distances[each][1] == 'M':
            m += 1
        else:
            f += 1

    if m > f:
        return 'M'
    else:
        return 'F'


# Test all 6 possibilities
for k in [1, 3]:
    print("k =", k)
    print("Euclidean:", knn(point, dataset, k, euclideanDistance))
    print("Manhattan:", knn(point, dataset, k, manhattanDistance))
    print("Hamming:", knn(point, dataset, k, hammingDistance))
```

Precision, Recall and F-measure:

Accuracy alone doesn't tell the whole story, so for each of the 9 results above you must also report the precision, recall, and f-measure.

Compute them separately for each tag in the dataset. For a given tag T:

Precision(T) = (number of test vectors correctly classified as T) / (number of test vectors your model classified as T)

Recall(T) = (number of test vectors correctly classified as T) / (number of test vectors whose true tag is T)

F-measure(T) = 2 * Precision(T) * Recall(T) / (Precision(T) + Recall(T))

Once you are done, each of your 9 results should include accuracy, precision, recall, and f-measure.

Then answer: Is the combination of k and distance function that gave the best accuracy also the best for precision, recall, and f-measure? If not, explain the difference.

```python
#Add code here
for k in [1, 7, 15]:
    for name, function in [
        ("Euclidean", euclideanDistance),
        ("Manhattan", manhattanDistance),
        ("Hamming", hammingDistance)
    ]:
        correct = 0
        predictions = []

        for each in test_data:
            prediction = knn(each, train_data, k, function)
            predictions.append(prediction)

            if prediction == each[-1]:
                correct += 1
        accuracy = correct / len(test_data)
        print("\n", name, "k =", k, "Accuracy =", accuracy)
        for label in ['M', 'F']:
            tp = 0
            fp = 0
            fn = 0
            for i in range(len(test_data)):
                actual = test_data[i][-1]
                predicted = predictions[i]

                if predicted == label and actual == label:
                    tp += 1
                elif predicted == label and actual != label:
                    fp += 1
                elif predicted != label and actual == label:
                    fn += 1

            precision = tp / (tp + fp) if tp + fp > 0 else 0
            recall = tp / (tp + fn) if tp + fn > 0 else 0
            fmeasure = 2 * precision * recall / (precision + recall) if precision + recall > 0 else 0

            print(label, "Precision =", round(precision, 3),
                  "Recall =", round(recall, 3),
                  "F-measure =", round(fmeasure, 3))

    # For each combination, I stored the predictions and compared them with the actual labels.
    # I counted the true positives, false positives, and false negatives for M and F separately, then used those counts to calculate precision, recall, and F-measure.
```
